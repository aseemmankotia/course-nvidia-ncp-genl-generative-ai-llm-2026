# Chapter 12 Lab — Timed Practice Exam Harness & Weak-Domain Review

**Estimated time:** 120 minutes · **Read-and-run.** The 65-question set lives in
`../practice-tests/`. This harness scores your answers and tags weak domains so your review is
evidence-based, not vibes-based.

```python
# Paste your answers as {question_number: 'A'/'B'/'C'/'D'}. Example key below is illustrative.
answer_key = {1:'B',2:'C',3:'A',4:'D',5:'B',6:'A',7:'C',8:'B',9:'D',10:'A'}
domains = {1:'Architecture',2:'Architecture',3:'Prompting',4:'RAG',5:'Data',
           6:'Customization',7:'Distributed',8:'Inference',9:'Serving',10:'MLOps'}
my_answers = {1:'B',2:'A',3:'A',4:'D',5:'B',6:'C',7:'C',8:'B',9:'A',10:'A'}  # <- edit
```

```python
import collections
correct=sum(1 for q,a in my_answers.items() if answer_key.get(q)==a)
total=len(answer_key)
miss=collections.Counter(domains[q] for q in answer_key if my_answers.get(q)!=answer_key[q])
print(f'Score: {correct}/{total} = {100*correct/total:.0f}%  (NCP-GENL passing is ~70%)')
print('Weakest domains (most missed):')
for dom,c in miss.most_common(): print(f'  {dom}: {c} missed')
print('\nPacing: 65 questions / 120 min ~ 1.8 min each. Flag-and-return rather than stalling.')
```

## Key takeaway
Passing the timed exam is about **disciplined pacing** and **evidence-based review** of weak
domains — not re-reading everything. Use the score breakdown to target your last study hours.
